```python
# Import the libraries
from llama_index.llms.ollama import Ollama
import pytesseract
from pdf2image import convert_from_path
import glob

# Set the path to the Tesseract executable
pytesseract.pytesseract.tesseract_cmd = r"C:\Program Files\Tesseract-OCR\tesseract.exe"

# Create an instance of the OLLAMA model
llm = Ollama(model="llama3", request_timeout=480.0)

# Convert the PDF image to text
pdfs = glob.glob(r"./scansmpl.pdf")
print(pdfs)
context = ""

for pdf_path in pdfs:
    print(f"Processing {pdf_path}...")
    pages = convert_from_path(pdf_path, 500)
    print(f"Pages: {pages}")

    for pageNum,imgBlob in enumerate(pages):
        text = pytesseract.image_to_string(imgBlob,lang='eng')
        context += text + "\n"
        print(context)

# Read and ask the document
question = "Summarize the document"
resp = llm.complete(f"{question}:\n{context}")
print(resp.text)
```
