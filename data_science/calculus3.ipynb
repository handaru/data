```python
import numpy as np
import matplotlib.pyplot as plt

# Define the function C(x, y)
def C(x, y):
    return x**2 + 2*y**2 - 4*x - 6*y + x*y + 100

# Create a grid of x and y values
x = np.linspace(-10, 10, 100)
y = np.linspace(-10, 10, 100)
X, Y = np.meshgrid(x, y)
Z = C(X, Y)

# Plot the surface
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')
surf = ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='none')

# Add labels and title
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('C(x, y)')
ax.set_title('3D Visualization of C(x, y)')

# Add color bar
fig.colorbar(surf, shrink=0.5, aspect=5)

plt.show()
```

```python
!pip install sympy
from sympy import symbols, Eq, solve

# Step 1: Define the variables
x, y = symbols('x y')

# Step 2: Define the partial derivative equations
# These are the equations that derived:
# ∂C/∂x = 2x - 4 + y = 0
# ∂C/∂y = 4y - 6 + x = 0
eq1 = Eq(2*x - 4 + y, 0)
eq2 = Eq(4*y - 6 + x, 0)

# Step 3: Solve the system of equations
# The solve function returns a dictionary of solutions
solution = solve((eq1, eq2), (x, y))

# Step 4: Print the solution
print("The system of equations is:")
print(f"Equation 1: {eq1}")
print(f"Equation 2: {eq2}")
print("\nSolution for (x, y):")
print(solution)

# Access individual values
x_critical = solution[x]
y_critical = solution[y]

print(f"\nCritical point:")
print(f"x = {x_critical}")
print(f"y = {y_critical}")

# To verify, substitute these values back into the original partial derivatives
print("\nVerification:")
print(f"∂C/∂x at critical point: {2*x_critical - 4 + y_critical}")
print(f"∂C/∂y at critical point: {4*y_critical - 6 + x_critical}")

#  Calculate the Minimum Cost
MC = (10/7)**2 + 2*(8/7)**2 - 4*(10/7) - 6*(8/7) + (10/7)*(8/7) + 100
print(f"Minimum cost: {MC}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# 1. Define the Cost Function
def cost_function(x, y):
    return x**2 + 2*y**2 - 4*x - 6*y + x*y + 100

# 2. Generate Data for the 3D Surface Plot
# Create a meshgrid for x and y values
x_range = np.linspace(-5, 10, 100) # Range for raw material
y_range = np.linspace(-5, 10, 100) # Range for labor hours
X, Y = np.meshgrid(x_range, y_range)

# Calculate the Z (cost) values for each (x, y) pair
Z = cost_function(X, Y)

# 3. Define the Critical Point (Minimum Cost)
x_critical = 10/7
y_critical = 8/7
z_critical = cost_function(x_critical, y_critical)

# 4. Create the 3D Plot
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Plot the surface
surface = ax.plot_surface(X, Y, Z, cmap='viridis', alpha=0.8, rstride=100, cstride=100)

# Plot the critical point (minimum)
ax.scatter(x_critical, y_critical, z_critical, color='red', s=100, label=f'Minimum Cost\n({x_critical:.2f}, {y_critical:.2f}, {z_critical:.2f})')

# 5. Customize the Plot
ax.set_xlabel('Raw Material (x)')
ax.set_ylabel('Labor Hours (y)')
ax.set_zlabel('Cost C(x, y)')
ax.set_title('Cost Function Surface with Minimum Point')
ax.legend()

# Add a color bar
fig.colorbar(surface, shrink=0.5, aspect=5, label='Cost C(x, y)')

# Add contour lines on the x-y plane to show the level sets
ax.contour(X, Y, Z, levels=np.linspace(Z.min(), Z.max(), 20), cmap='viridis_r', offset=Z.min())

plt.tight_layout()
plt.show()

# Optional: Generate a 2D contour plot (top-down view)
plt.figure(figsize=(8, 7))
contour = plt.contourf(X, Y, Z, levels=50, cmap='viridis')
plt.colorbar(contour, label='Cost C(x, y)')
plt.scatter(x_critical, y_critical, color='red', marker='o', s=100, edgecolor='black', zorder=5,
            label=f'Minimum Cost ({x_critical:.2f}, {y_critical:.2f})')
plt.xlabel('Raw Material (x)')
plt.ylabel('Labor Hours (y)')
plt.title('Cost Function Contour Plot')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Define the range for x
x_vals = np.linspace(-10, 10, 400)

# Define the equations as functions of x
# From eq1: 2x - 4 + y = 0 → y = -2x + 4
# From eq2: 4y - 6 + x = 0 → y = (6 - x)/4
y1 = -2 * x_vals + 4
y2 = (6 - x_vals) / 4

# Plot the lines
plt.figure(figsize=(8, 6))
plt.plot(x_vals, y1, label='eq1: y = -2x + 4', color='blue')
plt.plot(x_vals, y2, label='eq2: y = (6 - x)/4', color='red')

# Highlight the intersection point
from sympy import symbols, Eq, solve
x, y = symbols('x y')
eq1 = Eq(2*x - 4 + y, 0)
eq2 = Eq(4*y - 6 + x, 0)
solution = solve((eq1, eq2), (x, y))
x_sol, y_sol = float(solution[x]), float(solution[y])
plt.plot(x_sol, y_sol, 'go', label=f'Intersection: ({x_sol:.2f}, {y_sol:.2f})')

# Labels and legend
plt.xlabel('x')
plt.ylabel('y')
plt.title('Visualization of eq1 and eq2')
plt.grid(True)
plt.legend()
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.show()
```
