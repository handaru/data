```python
import pandas as pd
df = pd.DataFrame({'key': ['K0', 'K1', 'K2', 'K3', 'K4', 'K5'],
                    'A': ['A0', 'A1', 'A2', 'A3', 'A4', 'A5'],
                    'B': ['B0', 'B1', 'B2', 'B3', 'B4', 'B5']})
print(df.index.values)
print(df.shape)
df
```

```python
other = pd.DataFrame({'key': ['K0', 'K1', 'K2'],
                       'B': ['B0', 'B1', 'B2'],
                       'C': ['C0', 'C1', 'C2'],
                       'D': ['D0', 'D1', 'D2']})
print(other.index.values)
print(other.shape)
other
```

```python
result = df.join(other, lsuffix='_caller', rsuffix='_other')
result
```

```python
import numpy as np

def matrix_join(left_array, right_array, left_indices, right_indices):
    
    # Create index mapping 
    index_map = left_indices[:, np.newaxis] == right_indices[np.newaxis, :]
    print("Index Map:\n", index_map)
    
    # Initialize result array with NaNs
    result = np.full((left_array.shape[0], left_array.shape[1] + right_array.shape[1]), 
                     np.nan, dtype=object)
    print("Initial Result Array:\n", result)
    
    # Copy left array
    result[:, :left_array.shape[1]] = left_array
    print("After Copying Left Array:\n", result)
    
    # Use matrix operations to fill right part
    for i in range(len(left_indices)):
        matches = np.where(index_map[i])[0]
        if len(matches) > 0:
            result[i, left_array.shape[1]:] = right_array[matches[0]]
            print(f"After Filling Row {i}:\n", result)
    
    return result
```

```python
# Extract values and indices

left_arr = df.values
right_arr = other.values
left_idx = df.index.values
right_idx = other.index.values
```

```python
matrix_join_result = matrix_join(left_arr, right_arr, left_idx, right_idx)
matrix_join_result
```

```python
# Convert back to DataFrame for display
result_df = pd.DataFrame(
    matrix_join_result,
    columns=list(df.columns) + list(other.columns),
    index=df.index
)
result_df
```
