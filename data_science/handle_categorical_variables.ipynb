```python
import pandas as pd

path ="../datasets/Iris.csv"
df = pd.read_csv(path)

df.columns
```

```python
df["Species"].unique().tolist()
```

###  Label Binarizer

```python
from sklearn.preprocessing import LabelBinarizer

# creating a copy of the
# original data frame
df1 = df.copy().drop(columns=['Id'])

# creating an object 
# of the Label Binarizer
label_binarizer = LabelBinarizer()

# fitting the column 
# Species to LabelBinarizer
label_binarizer_output = label_binarizer.fit_transform( df1['Species'] )

# creating a data frame from the object
result_df = pd.DataFrame(label_binarizer_output,
                         columns = label_binarizer.classes_)

df1 = pd.concat([df1, result_df], axis = 1).drop(columns=['Species'])
display(df1)
```

```python
df1.columns.tolist()
```

### Binary Encoding

```python
%pip install category_encoders
import category_encoders as cat_encoder

# creating a copy of the
# original data frame
df2 = df.copy().drop(columns=['Id'])

# creating an object BinaryEncoder
encoder = cat_encoder.BinaryEncoder(cols = ['Species'])

# fitting the columns to a data frame
result_df = encoder.fit_transform(df2['Species'])

df2 = pd.concat([df2, result_df], axis = 1).drop(columns=['Species'])
display(df2)
```

```python
df2.columns.tolist()
```

### One Hot Encoding

```python
# creating a copy of the
# original data frame
df3 = df.copy().drop(columns=['Id'])

df3 = pd.get_dummies(df3,columns = ['Species'])

display(df3)
```

```python
df3.columns.tolist()
```

### Label Encoding

```python
# creating a copy of the
# original data frame
df4 = df.copy().drop(columns=['Id'])

# Convert target variable (Species) as category
df4.Species = df4.Species.astype('category')

df4["Species"] = df4.Species.cat.codes
display(df4)
```

```python
df4.columns.tolist()
```
