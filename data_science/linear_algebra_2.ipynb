```python
import numpy as np

# Define three vectors and the zero vector using NumPy arrays
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
c = np.array([7, 8, 9])
zero_vector = np.array([0, 0, 0])

print(f"Vector a: {a}")
print(f"Vector b: {b}")
print(f"Vector c: {c}")
print(f"Zero vector: {zero_vector}\n")

# --- Commutativity: a + b = b + a ---
print("--- Commutativity: a + b = b + a ---")
print(f"a + b: {a + b}")
print(f"b + a: {b + a}")
print(f"Is a + b equal to b + a? {np.array_equal(a + b, b + a)}\n")

# --- Associativity: (a + b) + c = a + (b + c) ---
print("--- Associativity: (a + b) + c = a + (b + c) ---")
print(f"(a + b) + c: {(a + b) + c}")
print(f"a + (b + c): {a + (b + c)}")
print(f"Is (a + b) + c equal to a + (b + c)? {np.array_equal((a + b) + c, a + (b + c))}\n")

# --- Adding the zero vector: a + 0 = a ---
print("--- Adding the zero vector: a + 0 = a ---")
print(f"a + zero_vector: {a + zero_vector}")
print(f"Is a + zero_vector equal to a? {np.array_equal(a + zero_vector, a)}\n")

# --- Subtracting a vector from itself: a - a = 0 ---
print("--- Subtracting a vector from itself: a - a = 0 ---")
print(f"a - a: {a - a}")
print(f"Is a - a equal to the zero vector? {np.array_equal(a - a, zero_vector)}\n")

# Verification: The zero vector returned by numpy.subtract is also the zero vector we defined
print("--- Verification: np.subtract(a, a) vs zero_vector ---")
print(f"np.subtract(a, a): {np.subtract(a, a)}")
print(f"Is np.subtract(a, a) equal to the zero vector? {np.array_equal(np.subtract(a, a), zero_vector)}")
```
