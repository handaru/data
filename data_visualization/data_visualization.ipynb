```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv('../datasets/Iris.csv')

sns.stripplot(x="Species", y="PetalWidthCm", data=df)

# Set plot title and axis labels
plt.title("Iris Flowers Classification based-on Petal Width Dimension")
plt.xlabel("Species")
plt.ylabel("PetalWidthCm")

plt.show()
```

```python
import seaborn as sns
from matplotlib import pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from matplotlib.colors import ListedColormap

x = df["PetalWidthCm"]
y = df["PetalLengthCm"]
z = df["Species"].astype('category').cat.codes

# axes instance
fig = plt.figure(figsize=(6,6))
ax = Axes3D(fig, auto_add_to_figure=False)
fig.add_axes(ax)

# plot
sc = ax.scatter(x, y, z, s=40, marker='o', alpha=1)
ax.set_xlabel('PetalWidthCm')
ax.set_ylabel('PetalLengthCm')
ax.set_zlabel('Species')

ax.set_zticks([0, 1, 2]) # type: ignore
ax.set_zticklabels(['Setosa', 'Versicolor', 'Virginica'], rotation=45, fontsize=8) # type: ignore

# save
plt.savefig("scatter_hue", bbox_inches='tight')
```
