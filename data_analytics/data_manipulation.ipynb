```python
states_around_mojave = ['California', 'Arizona', 'Nevada', 'Utah']

homelessness[['state','individuals','family_members','state_pop']][homelessness["state"].isin(states_around_mojave)].sort_values('state_pop', ascending=False)
```

```python
homelessness['total'] = homelessness['individuals'] + homelessness['family_members']

homelessness[['state','individuals','family_members','total','state_pop']][homelessness['state'].isin(states_around_mojave)].sort_values('state_pop', ascending=False).head()
```

```python
def iqr(column):
    return column.quantile(0.75) - column.quantile(0.25)
    
print(iqr(homelessness["total"]))
print(homelessness["total"].agg(iqr))
```

```python
print(iqr(homelessness["state_pop"]))
print(homelessness["state_pop"].agg(iqr))
```
