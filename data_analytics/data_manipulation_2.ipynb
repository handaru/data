## loc

loc (Label-based indexing)<br>
loc is primarily label-based, meaning use the actual row and column labels (names) to select data.<br><br>
Row Selection: Provide the label(s) of the row(s) that we want.
- Column Selection: Provide the label(s) of the column(s) that we want.
- Slicing: When slicing with loc, both the start and stop labels are inclusive.

```python
# Create a DataFrame
import pandas as pd

data = {'A': [1, 2, 3], 'B': [4, 5, 6], 'C': [7, 8, 9]}
df = pd.DataFrame(data, index=['row1', 'row2', 'row3'])
print(df)
```

```python
data2 = data
```

```python
df2 =  pd.DataFrame(data2)
df2
```

```python
# Select a single row by label
df.loc['row2']
```

```python
df.loc[['row2']]
```

```python
df.loc[['row1','row2', 'row3']]
```

```python
# Select multiple rows by labels
df.loc[['row1', 'row3']]
```

```python
# Select a single cell by row and column labels
df.loc['row1', 'A']
```

```python
# Select a range of rows and columns (inclusive slicing)
df.loc['row1':'row2', 'A':'C']
```

```python
work_df = df.loc[:,['A','C']]
work_df
```

```python
df.loc[:,'A']
```

```python
# Conditional selection with loc
df.loc[df['A'] > 1, ['A','C']]
```

```python
df.loc[df['A'] > 1, 'A':'C']
```

## iloc

iloc (Integer-location based indexing)<br>
iloc is primarily integer-location based, meaning use the integer positions (from 0 to n-1) of rows and columns to select data.<br><br>
- Row Selection: Provide the integer position(s) of the row(s) taht we want.
- Column Selection: Provide the integer position(s) of the column(s) that we want.
- Slicing: When slicing with iloc, the start position is inclusive, and the stop position is exclusive (just like standard Python list slicing).

```python
# Select a single row by integer position
df.iloc[1] # Selects the second row (index 1)
```

```python
df.iloc[[1]]
```

```python
# Select multiple rows by integer positions
df.iloc[[0, 2]] # Selects the first and third rows
```

```python
# Select a single cell by integer row and column positions
df.iloc[0, 1] # Selects the cell at row 0, column 1
```

```python
df.loc['row1', 'B']
```

```python
# Select a range of rows and columns (exclusive slicing for stop)
df.iloc[0:2, 0:2] # Selects rows 0 and 1, columns 0 and 1
```

## When to use which?

*   **Use `loc`** when we care about the actual names of our rows and columns. This is often the case when our DataFrame has meaningful labels (e.g., dates as index, specific sensor names as columns).
*   **Use `iloc`** when we need to access data by its position in the DataFrame, regardless of its labels. This is common when iterating through rows, selecting the first/last 'n' rows, or when the row/column labels are not unique or meaningful for our current task.

## Reset Index

If we reset the index of a DataFrame, the default integer index (0, 1, 2...) becomes a column, and we get a new default integer index.

```python
df_reset = df.reset_index()
print(df_reset)
```
